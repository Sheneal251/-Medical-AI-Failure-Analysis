Import Require Library

```python
import torch
from transformers import pipeline
```

Load pretrained owned model

```python
model = pipeline (
    "text-generation",
    model="Qwen/Qwen2.5-1.5B-Instruct",
    device=0 )
```

Evulate 30 medical Questions


```python
answers=[]
for question in sample["question"]:

    result = model(
    [{"role":"user","content": question}],
    max_new_tokens=150,
    do_sample=False
)
    answers.append(result[0]["generated_text"][-1]["content"])

sample = sample.copy()
sample["ai_answer"] = answers
sample[["question","answer", "ai_answer"]]
```

Measure Failure 2: Medical Procedure Accuracy

```python
questions = [
    "What procedure removes fluid from the pericardial sac?",
    "What procedure removes fluid from the pleural space?",
    "What procedure removes fluid from the abdomen?",
    "A patient has cardiac tamponade. What procedure is needed?",
    "A patient has pleural effusion. What procedure removes the fluid?"
]

for question in questions:
    result = model(
        [{"role": "user", "content": question}],
        max_new_tokens=100,
        do_sample=False
    )
    print(question)
    print(result[0]["generated_text"][-1]["content"])
```

Test 5 variations

```python
questions = [
    "How should cardiac tamponade be treated?",
    "A patient has fluid compressing the heart. What treatment is needed?",
    "What is the emergency treatment for cardiac tamponade?",
    "A patient has fluid in the pericardial sac causing low blood pressure. What should doctors do?",
    "How do doctors remove fluid around the heart during cardiac tamponade?"
]

for question in questions:
    result = model(
        [{"role": "user", "content": question}],
        max_new_tokens=150,
        do_sample=False
    )
    print(question)
    print(result[0]["generated_text"][-1]["content"])
```

Tested Qwen2.5-1.5B-Instruct using five variations of cardiac tamponade questions. The model correctly identified pericardiocentesis in one response, provided partially correct answers in two, and produced incorrect or potentially unsafe recommendations in two. Results suggest that medical response accuracy can vary depending on question phrasing.
